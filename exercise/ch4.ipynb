```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

x = np.linspace(-3.5, 3.5, 200)
print(x)

MU = 0
SIGMA = 1
px = stats.norm.pdf(x, loc=MU, scale=SIGMA)

plt.plot(x, px)
plt.title('Normal Distribution')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.grid()
plt.show()
```

```python
x = np.linspace(0, 10, 200)

MU = 0
SIGMA = 1
LOC =0
px = stats.lognorm.pdf(x, s=SIGMA, loc=LOC, scale=np.exp(MU))

plt.plot(x, px)
plt.title('Log-Normal Distribution')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.grid()
plt.show()
```

```python
x = np.linspace(0, 10, 200)

DF = 4
px = stats.chi2.pdf(x, df=DF)

plt.plot(x, px)
plt.title('Chi-Squared Distribution')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 200)

DF = 2
SCALE = 1
LOC = 0
px = stats.t.pdf(x, df=DF, loc=LOC, scale=SCALE)

plt.plot(x, px)
plt.title('Student\'s t-qDistribution')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.grid()
plt.show()
```

```python
x = stats.norm.rvs(loc=4, scale=2, size=1000)

params = stats.norm.fit(x)
loc = params[0]
scale = params[1]

print(f'mu={loc}, sigma={scale}')

plt.hist(x, color='#bbbbbb', bins='sturges', density=True)

xline = np.linspace(-2, 10, 200)
estimated_dist = stats.norm.pdf(xline, loc=loc, scale=scale)
plt.plot(xline, estimated_dist, color='#111111')
plt.show()
```

```python
x = stats.chi2.rvs(df=4, size=1000)
params = stats.chi2.fit(x, floc=0, fscale=1)
df = params[0]
loc = params[1]
scale = params[2]

print(f'{df=}, {loc=}, {scale=}')

plt.hist(x, color='#bbbbbb', bins='sturges', density=True)
xline = np.linspace(-2, 20, 200)
estimated_dist = stats.chi2.pdf(xline, df=df, loc=loc, scale=scale)
plt.plot(xline, estimated_dist, color='#111111')
plt.xlim(left=0)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))
df = pd.read_csv('./datasets/ch2_dataset_train2.csv')

df_dropna = df.dropna(subset=['temp1'])
df_normal = df_dropna[df_dropna['label'] =='normal']
axes[0].hist(df_normal['temp1'], bins='sturges', color='#bbbbbb', edgecolor='#111111', density=True)

params = stats.norm.fit(df_normal['temp1'])
mu = params[0]
sigma = params[1]
print(f'mu={mu}, sigma={sigma}')

xline = np.linspace(df_normal['temp1'].min(), df_normal['temp1'].max(), 200)
estimated_dist = stats.norm.pdf(xline, loc=mu, scale=sigma)
axes[0].plot(xline, estimated_dist, color='red', label='Estimated Normal PDF')

axes[0].set_title('Histogram of temp1')
axes[0].legend()
axes[0].set_xlabel('temp1')
axes[0].set_ylabel('Density')
axes[0].grid()


x_train = df_normal['temp1'].to_numpy()

stats.probplot(x_train, dist='norm', plot=axes[1])
axes[1].set_title('Q-Q Plot for temp1')
axes[1].set_xlabel('Theoretical Quantiles')
axes[1].set_ylabel('Ordered Values')
axes[1].grid()
plt.show()
```

```python
import numpy as np

# Shapiro-Wilk Test
stat, p = stats.shapiro(x_train)
print(f'Shapiro-Wilk Test: stat={stat}, p={p}')

# Kolmogorov-Smirnov Test
mean= np.mean(x_train)
std = np.std(x_train, ddof=1)
stat, p = stats.kstest(x_train, stats.norm.cdf, args=(mean, std))
print(f'Kolmogorov-Smirnov Test: stat={stat}, p={p}')
```

```python
import pandas as pd
import numpy as np
from scipy import stats

import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('./datasets/ch2_dataset_train2.csv')

# remvoe rows with missing values in 'temp1'
df_dropna = df.dropna(subset=['temp1'])

# extract 'temp1' values for normal samples
df_normal = df_dropna[df_dropna['label'] == 'normal']

# convert to numpy array
x_train = df_normal['temp1'].to_numpy()

# Step 1: Fit a normal distribution to the data
params = stats.norm.fit(x_train)
mu = params[0]
sigma = params[1]

# Step 2:

# Step 3: 
TARGET_FP_RATE = 0.0027 # 0.27% false positive rate corresponds to 3-sigma in a normal distribution
a_th = stats.chi2.ppf(1 - TARGET_FP_RATE, df=1) # chi-squared threshold for 1 degree of freedom

# Print all paramters
print(f'mu={mu}, sigma={sigma}, a_th={a_th}')

MU = mu
SIGMA = sigma
A_TH = a_th

df_inference = pd.read_csv('./datasets/ch2_dataset_inference2.csv')

# remove rows with missing values in 'temp1'
df_inference_dropna = df_inference.dropna(subset=['temp1'])
# extract 'temp1' values for inference samples
x_inference = df_inference_dropna['temp1'].to_numpy()

# Step 4: Calculate the anomaly score for each inference sample
anomaly_scores = ((x_inference - MU) / SIGMA) ** 2

# Step 5: Classify samples as anomalous if their anomaly score exceeds the threshold
pred = np.where(anomaly_scores > A_TH, 'anomaly', 'normal')

print(pred)

# 異常度をDataFrameに列として追加
df_inference_dropna['anomaly_score'] = anomaly_scores
# 描画用のFigureとAxesを生成
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(6, 4))

# 正常データのヒストグラム描画
df_inference_norm = df_inference_dropna[df_inference_dropna['label'] == 'normal']
sns.histplot(data=df_inference_norm['anomaly_score'], binwidth=1,
             color='#bbbbbb', ax=ax,
             stat='density', label='normal')
# 異常データのヒストグラム描画
df_inference_anom = df_inference_dropna[df_inference_dropna['label'] == 'anomaly']
sns.histplot(data=df_inference_anom['anomaly_score'], binwidth=1,
             color="#FF0000", ax=ax,
             stat='density', label='anomaly')
# しきい値の線を描画
ax.vlines(A_TH, 0, 0.3, color='black', lw=2)
ax.text(A_TH, 0.3, f'Threshold={round(A_TH, 4)}',
        verticalalignment='bottom', horizontalalignment='center')
# 凡例を追加
ax.legend()
# グラフを表示
plt.show()
```
